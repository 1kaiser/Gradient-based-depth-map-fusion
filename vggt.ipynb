```python
%cd /content/vggt/

from vggt.models.vggt import VGGT
from vggt.utils.load_fn import load_and_preprocess_images
from vggt.utils.pose_enc import pose_encoding_to_extri_intri
from vggt.utils.geometry import unproject_depth_map_to_point_map

import torch


device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.bfloat16
model = VGGT().to(device)
# model.load_state_dict(torch.hub.load_state_dict_from_url("https://huggingface.co/facebook/VGGT-1B/resolve/main/model.pt"))

# Step 3: Apply the state_dict to the model
model.load_state_dict(torch.load("/content/model.pt", map_location=device))
```

```python
image_paths = [
'/content/input_depthMaps/input_frames/frame_0001.png',
'/content/input_depthMaps/input_frames/frame_0005.png',
'/content/input_depthMaps/input_frames/frame_0010.png',
# '/content/input_depthMaps/input_frames/frame_0015.png',
# '/content/input_depthMaps/input_frames/frame_0021.png',
# '/content/input_depthMaps/input_frames/frame_0030.png',
# '/content/input_depthMaps/input_frames/frame_0040.png',
# '/content/input_depthMaps/input_frames/frame_0050.png',
# '/content/input_depthMaps/input_frames/frame_0070.png',
]

images = load_and_preprocess_images(image_paths).to(device)[None]

with torch.no_grad(), torch.cuda.amp.autocast(dtype=dtype):
    tokens, ps_idx = model.aggregator(images)
    pose_enc = model.camera_head(tokens)[-1]
    extrinsic, intrinsic = pose_encoding_to_extri_intri(pose_enc, images.shape[-2:])
    depth_map, depth_conf = model.depth_head(tokens, images, ps_idx)
    point_map = unproject_depth_map_to_point_map(depth_map.squeeze(0), extrinsic.squeeze(0), intrinsic.squeeze(0))
    query = torch.FloatTensor([[100., 200.], [60.72, 259.94]]).to(device)
    track, vis, conf = model.track_head(tokens, images, ps_idx, query_points=query[None])

depth_np = depth_map.detach().cpu().numpy()[0, 0]
depth_norm = (depth_np - depth_np.min()) / (depth_np.max() - depth_np.min() + 1e-8)

import matplotlib.pyplot as plt
plt.imshow(depth_norm, cmap='plasma'); plt.colorbar(shrink=0.4); plt.axis('off')
plt.title("Normalized Depth Map"); plt.show()
```

### point clouds frames converstion to gb using trimesh library

```python
!uv pip install trimesh==4.7.0 scipy==1.15.3 matplotlib==3.10.0
```

```python
import numpy as np
import trimesh
from scipy.spatial.transform import Rotation
import torch

# 🔁 Safe conversion
def to_numpy(x):
    if isinstance(x, torch.Tensor):
        return x.detach().cpu().numpy()
    return x

# 🔢 Convert model outputs
point_map_np = to_numpy(point_map)                     # (9, 392, 518, 3)
images_np    = to_numpy(images[0].permute(0, 2, 3, 1)) # (9, 392, 518, 3)
extrinsic_np = to_numpy(extrinsic[0])                  # (9, 3, 4)
conf_np      = to_numpy(depth_conf[0])                 # (9, 392, 518)

# 🧱 Prepare extrinsic 4×4 matrices
extrinsic_full = np.zeros((extrinsic_np.shape[0], 4, 4))
extrinsic_full[:, :3, :] = extrinsic_np
extrinsic_full[:, 3, 3] = 1

# 🔃 Flatten point cloud
vertices = point_map_np.reshape(-1, 3)
colors_rgb = (images_np.reshape(-1, 3) * 255).clip(0, 255).astype(np.uint8)
conf = conf_np.reshape(-1)

# 🎯 Filter by confidence
conf_thres = 50.0
conf_mask = conf >= np.percentile(conf, conf_thres)
vertices = vertices[conf_mask]
colors_rgb = colors_rgb[conf_mask]

# 🎨 Create scene
scene = trimesh.Scene()
scene.add_geometry(trimesh.points.PointCloud(vertices=vertices, colors=colors_rgb))

# 📷 Add cameras
for i in range(extrinsic_full.shape[0]):
    world_to_cam = extrinsic_full[i]
    cam_to_world = np.linalg.inv(world_to_cam)

    cam = trimesh.creation.cone(radius=0.02, height=0.05)
    cam.apply_transform(cam_to_world)
    cam.visual.face_colors = np.array([[255, 0, 0, 255]] * len(cam.faces))
    # scene.add_geometry(cam)

# 🧭 Align to first camera view
def get_opengl_matrix():
    mat = np.eye(4)
    mat[1, 1] = -1
    mat[2, 2] = -1
    return mat

align_rot = np.eye(4)
align_rot[:3, :3] = Rotation.from_euler("y", 180, degrees=True).as_matrix()
scene.apply_transform(np.linalg.inv(extrinsic_full[0]) @ get_opengl_matrix() @ align_rot)

# 📦 Export or preview
# scene.show()

_ = scene.export("/content/combined_scene.glb")
```
